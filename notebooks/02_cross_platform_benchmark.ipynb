# Cross-platform atlas adaptation

This notebook inspects the frozen human Visium HD → Xenium/CosMx results reported in the manuscript. The biological data are from SPATCH; the exact prediction-lock and evaluation workflows are preserved in the companion reproducibility repository.

```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd

root = Path.cwd()
if not (root / 'examples').is_dir():
    root = root.parent
evidence = root / 'examples' / 'manuscript_evidence' / 'figure_05'
xenium = pd.read_csv(evidence / '01_PRIMARY_RESULTS_TABLE.tsv', sep='\t')
cosmx = pd.read_csv(evidence / '03_PAIR_LEVEL_RESULTS.tsv', sep='\t')
```

```python
x = xenium[['technology', 'specimen', 'hidden_gene_count', 'IDW_median_spatial_pearson', 'MAP_median_spatial_pearson', 'IDW_log1p_RMSE', 'MAP_log1p_RMSE', 'MAP_pooled_residual_pearson']].copy()
c = cosmx.rename(columns={'target_platform':'technology', 'tumor':'specimen', 'IDW_median_gene_wise_spatial_pearson':'IDW_median_spatial_pearson', 'MAP_median_gene_wise_spatial_pearson':'MAP_median_spatial_pearson'})
c = c[['technology', 'specimen', 'hidden_gene_count', 'IDW_median_spatial_pearson', 'MAP_median_spatial_pearson', 'IDW_log1p_RMSE', 'MAP_log1p_RMSE', 'MAP_pooled_residual_pearson']]
results = pd.concat([x, c], ignore_index=True)
results
```

```python
fig, axes = plt.subplots(1, 2, figsize=(8, 3.2))
for i, row in results.iterrows():
    label = f"{row.technology}–{row.specimen}"
    axes[0].plot([0, 1], [row.IDW_median_spatial_pearson, row.MAP_median_spatial_pearson], marker='o', label=label)
    axes[1].plot([0, 1], [row.IDW_log1p_RMSE, row.MAP_log1p_RMSE], marker='o')
for ax, title in zip(axes, ['Spatial Pearson', 'log1p RMSE']):
    ax.set_xticks([0, 1], ['Registered IDW', 'AtlasTailor'])
    ax.set_title(title)
    ax.spines[['top', 'right']].set_visible(False)
axes[0].legend(frameon=False, fontsize=8, bbox_to_anchor=(1.02, -0.22), ncol=2)
fig.tight_layout()
```

The notebook reads locked result tables only. It does not recompute predictions or select examples using target outcomes.
