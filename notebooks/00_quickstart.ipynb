# Quickstart

Adapt a tiny target while target non-anchor columns remain inaccessible to prediction.

```python
from pathlib import Path
import tempfile
import hyperspatial as hs
from hyperspatial.io import read_reference, read_target
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
EX = ROOT / 'examples' / 'minimal_2d'
reference = read_reference(EX / 'reference.h5ad')
```

```python
panel = hs.design_panel(reference, budget=8, config=hs.DesignConfig(budget=8, budgets=(8,)))
panel.genes
```

```python
target = read_target(EX / 'target.h5ad', panel.genes)
result = hs.adapt(reference, target, panel.genes, config=hs.AdaptConfig(registration='pre_registered', seeds=(42,)))
```

```python
g = result.gene(result.genes[0])
(g.gene, g.support['selected_predictor'], g.registered_atlas.shape, g.adapted.shape)
```

```python
out = Path(tempfile.mkdtemp()) / 'adaptation'
result.export(out)
sorted(p.name for p in out.iterdir())
```
